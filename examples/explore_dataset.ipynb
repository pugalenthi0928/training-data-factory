# Explore Forge datasets

This notebook shows how to load and inspect JSONL datasets generated by Training Data Robo.

```python
from pathlib import Path

import pandas as pd

# Change this path to any dataset you created
path = Path("output/papers_rich_summary_qa_300.jsonl")
df = pd.read_json(path, lines=True)
df.head()
```

```python
# Basic stats
df["input_len"] = df["input_text"].str.len()
df["output_len"] = df["output_text"].str.len()
df[["input_len", "output_len"]].describe()
```

```python
# Per-task counts
df["task_name"].value_counts()
```
